```python
from e14c.web import web_search
results = await web_search('site:publishing.aip.org authors cover letter Review of Scientific Instruments related work cover letter', num_results=5)
for i, r in enumerate(results):
    print(i, r['title'], r['url'])
    print((r.get('snippet') or '')[:500], '\n')
```

```python
from e14c.web import fetch_parsed_page
url='https://publishing.aip.org/resources/researchers/author-instructions/'
page=await fetch_parsed_page(url)
for term in ['cover letter','related work','Review of Scientific Instruments']:
    print('\nTERM', term)
    idx=page.lower().find(term.lower())
    print(page[max(0,idx-1000):idx+1800] if idx>=0 else 'not found')
```

```python
import re
# Locate the journal-specific RSI section, not the table of contents.
for m in re.finditer(r'(?im)^### Review of Scientific Instruments.*$', page):
    print('match', m.start(), page[m.start():m.start()+3000])
```
